```python
import sys
sys.path.append('../src')

import pandas as pd
import matplotlib.pyplot as plt
from feature_engineering import calculate_invariant_mass, calculate_delta_r
```

```python
higgs = pd.read_parquet('../data/processed/higgs_processed.parquet')
qcd = pd.read_parquet('../data/processed/qcd_processed.parquet')
```

```python
higgs['myy'] = calculate_invariant_mass(
    higgs['pt_lead'], higgs['eta_lead'], higgs['phi_lead'],
    higgs['pt_sublead'], higgs['eta_sublead'], higgs['phi_sublead']
)

qcd['myy'] = calculate_invariant_mass(
    qcd['pt_lead'], qcd['eta_lead'], qcd['phi_lead'],
    qcd['pt_sublead'], qcd['eta_sublead'], qcd['phi_sublead']
)
```

```python
higgs['delta_r'] = calculate_delta_r(
    higgs['eta_lead'], higgs['phi_lead'],
    higgs['eta_sublead'], higgs['phi_sublead']
)

qcd['delta_r'] = calculate_delta_r(
    qcd['eta_lead'], qcd['phi_lead'],
    qcd['eta_sublead'], qcd['phi_sublead']
)    
```
