```python
import sys
sys.path.append('../src')

import pandas as pd
import matplotlib.pyplot as plt
from feature_engineering import calculate_invariant_mass, calculate_delta_r, calculate_delta_eta, calculate_delta_phi, calculate_pt_ratio, calculate_pt_asymmetry, calculate_pt_sum
```

```python
higgs = pd.read_parquet('../data/processed/higgs_processed.parquet')
qcd = pd.read_parquet('../data/processed/qcd_processed.parquet')
```

```python
higgs['delta_r'] = calculate_delta_r(
    higgs['eta_lead'], higgs['phi_lead'],
    higgs['eta_sublead'], higgs['phi_sublead']
)

qcd['delta_r'] = calculate_delta_r(
    qcd['eta_lead'], qcd['phi_lead'],
    qcd['eta_sublead'], qcd['phi_sublead']
)    
```

```python
higgs['delta_eta'] = calculate_delta_eta(higgs['eta_lead'], higgs['eta_sublead'])

qcd['delta_eta'] = calculate_delta_eta(qcd['eta_lead'], qcd['eta_sublead'])
```

```python
higgs['delta_phi'] = calculate_delta_phi(higgs['phi_lead'], higgs['phi_sublead'])

qcd['delta_phi'] = calculate_delta_phi(qcd['phi_lead'], qcd['phi_sublead'])
```

```python
higgs['pt_ratio'] = calculate_pt_ratio(higgs['pt_lead'], higgs['pt_sublead'])

qcd['pt_ratio'] = calculate_pt_ratio(qcd['pt_lead'], qcd['pt_sublead'])
```

```python
higgs['pt_asymmetry'] = calculate_pt_asymmetry(higgs['pt_lead'], higgs['pt_sublead'])

qcd['pt_asymmetry'] = calculate_pt_asymmetry(qcd['pt_lead'], qcd['pt_sublead'])
```

```python
higgs['pt_sum'] = calculate_pt_sum(higgs['pt_lead'], higgs['pt_sublead'])

qcd['pt_sum'] = calculate_pt_sum(qcd['pt_lead'], qcd['pt_sublead'])
```

```python
print(f"Higgs DataFrame after feature engineering:\n{higgs.head()}")
```

```python
print(f"QCD DataFrame after feature engineering:\n{qcd.head()}")
```

```python
higgs.to_parquet('../data/processed/higgs_with_features.parquet', index=False)
qcd.to_parquet('../data/processed/qcd_with_features.parquet', index=False)
```
